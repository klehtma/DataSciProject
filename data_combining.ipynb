```python
import pandas as pd
from pathlib import Path
```

```python
folder = Path("test_datasets")
files = list(folder.glob("*.csv"))
readable_files = [pd.read_csv(file) for file in files]
combined_dataset = pd.concat(readable_files, ignore_index = True)

output = Path("combined_dataset.csv")
combined_dataset.to_csv(output, index = False, encoding = "utf-8")
```
